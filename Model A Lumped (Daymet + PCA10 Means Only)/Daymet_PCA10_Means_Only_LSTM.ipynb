# Model A Lumped: Daymet + PCA10 means only

## Two input files
Keep this notebook and **daymet_pca10_means_only_discharge_1980_2009.csv** together in the same directory. Open the notebook there, select **Model A PyTorch (GPU)**, and run all cells in order. No original CAMELS, PCA, split, attribute, utility or other experiment files are required. Python needs NumPy, PyTorch, IPython and a Jupyter kernel. Outputs are generated under `outputs/`.

## Controlled experiment
Seed **42 only**. Exactly the original **1,326 training-donor basins and 100 unseen basins**. Training **1980–1999**, validation **2000–2004**, testing **2005–2009**, separately for seen and unseen catchments. The seven Daymet forcings are prcp, tmax, tmin, srad, vp, dayl and pet. The only static inputs are **PC01_mean through PC10_mean**. No subbasin standard-deviation descriptors, basin IDs, group labels, subbasin counts or split flags enter the model. Standard deviations calculated for normalization are scaling parameters, not extra input features.

The shared PCA basis and means are unchanged from the earlier PCA10 mean/std experiment; PCA is not refitted. One-layer LSTM, **64 hidden units**, linear discharge head, **365-day input windows**, Adam **0.001**, batch **128**, clipping **1**, **20 epochs**, **256 sampled windows per donor per epoch**, and at most **128 fixed validation windows per donor**. Best validation loss selects the checkpoint. There are **17 input channels and 21,313 parameters**; removing ten inputs necessarily changes the input-weight count while the hidden layer is unchanged. Original loss scaling, missing-data handling, window endpoints and prediction handling are preserved.

## Reuse and results
This notebook performs reading, preparation, training, validation, seen/unseen testing, independent metric verification and reporting. `outputs/basin_model_performance.csv` has one row per basin with Seen/Unseen, NSE, KGE, RMSE, MAE, bias, observation count and status. Daily predictions, checkpoints and preprocessing parameters are retained under `outputs/results_seed42/`. `outputs/model_performance_summary.csv` and `outputs/model_report.html` summarize results. Missing discharge stays missing; unavailable scores are not replaced by zero.

To retrain from scratch, copy these two input files into a new empty directory. Rerunning in the same directory resumes a matching checkpoint. GPU arithmetic can vary slightly across hardware/software despite a fixed seed. Earlier meteorology may supply window context for validation/testing; discharge is never a model input. The original geographical-overlap limitations of the basin split remain unchanged.


```python
from pathlib import Path
import csv, json, time, random, hashlib, base64, zlib, math, os
import numpy as np
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader, Sampler
from IPython.display import display, HTML
CSV_PATH = Path.cwd() / 'daymet_pca10_means_only_discharge_1980_2009.csv'
RUN_NAME = 'results_seed42'
OUTPUT_BASE = CSV_PATH.parent / 'outputs'
OUTPUT_BASE.mkdir(exist_ok=True)
OUT = OUTPUT_BASE / RUN_NAME
FEATURES = ['prcp', 'tmax', 'tmin', 'srad', 'vp', 'dayl', 'pet']
STATIC_FEATURES = ['PC01_mean', 'PC02_mean', 'PC03_mean', 'PC04_mean', 'PC05_mean', 'PC06_mean', 'PC07_mean', 'PC08_mean', 'PC09_mean', 'PC10_mean']
CFG = dict(seed=42, epochs=20, batch=128, windows_per_basin=256, validation_cap=128, lookback=365, hidden=64, learning_rate=0.001, gradient_clip=1.0, train_start='1980-01-01', validation_start='2000-01-01', test_start='2005-01-01', end_exclusive='2010-01-01', features=FEATURES, static_features=STATIC_FEATURES)
assert CSV_PATH.is_file(), 'Put the CSV beside this notebook, or edit CSV_PATH.'
assert torch.cuda.is_available(), 'Select a GPU Jupyter session/kernel.'
torch.set_num_threads(2)
random.seed(42)
np.random.seed(42)
torch.manual_seed(42)
torch.cuda.manual_seed_all(42)
torch.backends.cudnn.benchmark = False
DEVICE = torch.device('cuda:0')
print('GPU:', torch.cuda.get_device_name(0))
print('CSV:', CSV_PATH)
print('Only model inputs:', FEATURES)
print('Static inputs:', STATIC_FEATURES)
```

## 1. Read the combined dataset
Verify the column schema, original basin split, daily dates and unchanged metadata.


```python
EXPECTED_SPLIT_HASH="779bde7ce5db72c8158f26f161865c4987eb4ce6a0ac7469383256b41cc2d844"
def atomic(path,obj):
    tmp=path.with_suffix(path.suffix+'.tmp');tmp.write_text(json.dumps(obj,indent=2));tmp.replace(path)
def save_csv(path,rows,headers=None):
    headers=headers or list(rows[0])
    with open(path,'w',newline='') as f:
        w=csv.DictWriter(f,fieldnames=headers);w.writeheader();w.writerows(rows)

from itertools import groupby
records=[];preview=[];expected_dates=np.arange(np.datetime64('1980-01-01'),np.datetime64('2010-01-01'))
started=time.time()
with open(CSV_PATH,newline='') as f:
    reader=csv.reader(f);header=next(reader)
    expected=['basin_id','split','date',*FEATURES,'q_obs_mm_day',*STATIC_FEATURES,'seen_unseen','n_subbasins','single_subbasin_flag','time_split']
    assert header==expected,header
    for basin,group in groupby(reader,key=lambda row:row[0]):
        chunk=list(group);splits={r[1] for r in chunk};assert len(splits)==1
        split=next(iter(splits));dates=np.array([r[2] for r in chunk],dtype='datetime64[D]');assert np.array_equal(dates,expected_dates),basin
        data=np.array([r[3:11] for r in chunk],dtype='float32');assert data.shape==(len(dates),8)
        static=np.array(chunk[0][11:21],dtype='float64');assert np.isfinite(static).all()
        assert all(r[11:24]==chunk[0][11:24] for r in chunk),'Static feature or basin metadata changes within basin'
        seen='Seen' if split=='training_donor' else 'Unseen';assert chunk[0][21]==seen
        expected_period=np.where(dates<np.datetime64('2000-01-01'),'train',np.where(dates<np.datetime64('2005-01-01'),'validation','test'))
        assert np.array_equal(np.array([r[24] for r in chunk]),expected_period)
        assert np.all(data[np.isfinite(data[:,-1]),-1]>=0)
        records.append(dict(id=basin,split=split,dates=dates,x=data[:,:7].copy(),y=data[:,7].copy(),static=static,n_subbasins=int(chunk[0][22]),single_subbasin_flag=int(chunk[0][23])))
        if not preview:preview=chunk[:5]
        if len(records)%200==0:print('CSV catchments loaded:',len(records),flush=True)
ids=[r['id'] for r in records];assert len(ids)==len(set(ids))==1426
split_text=''.join(r['id']+'|'+r['split']+chr(10) for r in sorted(records,key=lambda r:r['id']))
assert hashlib.sha256(split_text.encode()).hexdigest()==EXPECTED_SPLIT_HASH,'Catchment split differs from original baseline.'
donors=[r for r in records if r['split']=='training_donor'];held=[r for r in records if r['split']=='unseen_test']
assert len(donors)==1326 and len(held)==100 and not {r['id'] for r in donors}&{r['id'] for r in held}
hash_csv=hashlib.sha256()
with open(CSV_PATH,'rb') as f:
    for part in iter(lambda:f.read(8*1024*1024),b''):hash_csv.update(part)
CSV_HASH=hash_csv.hexdigest();OUT.mkdir(exist_ok=True)
DIGEST=hashlib.sha256(json.dumps(dict(config=CFG,csv_sha256=CSV_HASH),sort_keys=True).encode()).hexdigest()
atomic(OUT/'data_audit.json',dict(csv=str(CSV_PATH),rows=sum(len(r['dates']) for r in records),basins=1426,donors=1326,unseen=100,csv_sha256=CSV_HASH,dynamic_inputs=FEATURES,static_inputs=STATIC_FEATURES,metadata_excluded=['basin_id','split','seen_unseen','n_subbasins','single_subbasin_flag','time_split'],loader_seconds=round(time.time()-started)))
display(HTML('<table border="1">'+''.join('<tr>'+''.join('<td>'+str(x)+'</td>' for x in row)+'</tr>' for row in [header]+preview)+'</table>'))
print('CSV is the sole data source. Exact original split verified.')
```

## 2. Prepare training and validation inputs
Fit scalers on donor training data only. Create balanced training samples and fixed validation windows.


```python
def fit_preprocessing(rs):
    assert len(rs)==1326 and all(r['split']=='training_donor' for r in rs)
    sx=np.zeros(7);sx2=np.zeros(7);nx=0;sy=0.;sy2=0.;ny=0;sigmas=[]
    for r in rs:
        train=r['dates']<np.datetime64(CFG['validation_start'])
        x=r['x'][train&np.isfinite(r['x']).all(1)].astype('float64');y=r['y'][train&np.isfinite(r['y'])].astype('float64')
        assert len(x)>0 and len(y)>0,r['id']
        sx+=x.sum(0);sx2+=(x*x).sum(0);nx+=len(x);sy+=y.sum();sy2+=(y*y).sum();ny+=len(y);sigmas.append(float(y.std()))
    p=dict(xmean=sx/nx,xstd=np.maximum(np.sqrt(np.maximum(sx2/nx-(sx/nx)**2,0)),1e-6),ymean=sy/ny,ystd=max(np.sqrt(max(sy2/ny-(sy/ny)**2,0)),1e-6))
    p['loss_scales']=np.maximum(np.array(sigmas)/p['ystd'],.1);return p
class Windows(Dataset):
    def __init__(self,rs,p,split,cap=None):
        self.rs,self.p,self.split=rs,p,split;self.x=[];self.indices=[]
        for r in rs:
            x=((r['x']-p['xmean'])/p['xstd']).astype('float32');self.x.append(x)
            d=r['dates'];L=CFG['lookback'];t=np.arange(L-1,len(d));bad=np.r_[0,np.cumsum(~np.isfinite(x).all(1))];keep=(bad[t+1]-bad[t+1-L])==0
            if split=='train':keep &= d[t]<np.datetime64(CFG['validation_start'])
            elif split=='validation':keep &= (d[t]>=np.datetime64(CFG['validation_start']))&(d[t]<np.datetime64(CFG['test_start']))
            elif split=='test':keep &= d[t]>=np.datetime64(CFG['test_start'])
            else:raise ValueError(split)
            if split!='test':keep &= np.isfinite(r['y'][t])
            ix=t[keep]
            if cap and len(ix)>cap:ix=ix[np.linspace(0,len(ix)-1,cap).astype(int)]
            self.indices.append(ix.astype('int32'))
        self.lengths=np.array([len(a) for a in self.indices]);self.ends=np.cumsum(self.lengths);self.starts=np.r_[0,self.ends[:-1]]
        if split=='train':assert (self.lengths>0).all()
    def __len__(self):return int(self.ends[-1])
    def __getitem__(self,i):
        bi=int(np.searchsorted(self.ends,i,side='right'));t=int(self.indices[bi][i-self.starts[bi]]);L=CFG['lookback']
        y=(self.rs[bi]['y'][t]-self.p['ymean'])/self.p['ystd'] if self.split!='test' else 0.
        dynamic=self.x[bi][t+1-L:t+1];static=np.broadcast_to(self.rs[bi]['static_scaled'],(L,len(STATIC_FEATURES)))
        inputs=np.concatenate([dynamic,static],axis=1)
        return torch.from_numpy(inputs),torch.tensor(y,dtype=torch.float32),bi,t
class EqualBasinSampler(Sampler):
    def __init__(self,ds,per_basin=256,seed=42):self.ds,self.per_basin,self.seed,self.epoch=ds,per_basin,seed,0
    def __len__(self):return len(self.ds.rs)*self.per_basin
    def __iter__(self):
        rng=np.random.default_rng(self.seed+self.epoch);bs=np.repeat(np.arange(len(self.ds.rs)),self.per_basin);rng.shuffle(bs)
        ix=self.ds.starts[bs]+rng.integers(self.ds.lengths[bs]);return iter(ix.tolist())
p=fit_preprocessing(donors)
static_train=np.stack([r['static'] for r in donors]);p['static_mean']=static_train.mean(0);p['static_std']=np.maximum(static_train.std(0),1e-6)
assert np.isfinite(p['static_mean']).all() and np.isfinite(p['static_std']).all()
for r in records:r['static_scaled']=((r['static']-p['static_mean'])/p['static_std']).astype('float32')

train=Windows(donors,p,'train');val=Windows(donors,p,'validation',cap=128)
sampler=EqualBasinSampler(train);loader=DataLoader(train,batch_size=128,sampler=sampler,num_workers=0,pin_memory=True)
val_loader=DataLoader(val,batch_size=256,num_workers=0,pin_memory=True)
assert len(sampler)==1326*256
# Changing test discharge cannot change test-window dates or model-input tensors.
r=held[0];altered=dict(r,y=np.full_like(r['y'],np.nan));ta=Windows([r],p,'test');tb=Windows([altered],p,'test')
assert np.array_equal(ta.indices[0],tb.indices[0]);assert torch.equal(ta[0][0],tb[0][0]);assert ta[0][1].item()==0.
np.savez(OUT/'preprocessing.npz',**p)
atomic(OUT/'fit_metadata.json',dict(fitted_ids=[r['id'] for r in donors],excluded_ids=[r['id'] for r in held],fit_end_exclusive='2000-01-01',static_features=STATIC_FEATURES,experiment_hash=DIGEST))
save_csv(OUT/'window_counts.csv',[dict(basin_id=r['id'],train_windows=int(n),validation_windows=int(v)) for r,n,v in zip(donors,train.lengths,val.lengths)])
print('Training windows:',len(train),'fixed validation windows:',len(val),'validation donors:',int((val.lengths>0).sum()))

assert train[0][0].shape==(365,17)
assert not {r['id'] for r in held}&{r['id'] for r in train.rs}
print('Static scaler fitted once per donor basin; 17 LSTM input channels.')
```

## 3. Define the LSTM, validation and checkpoint functions


```python
class MeteorologyLSTM(nn.Module):
    def __init__(self):
        super().__init__();self.lstm=nn.LSTM(17,64,batch_first=True);self.head=nn.Linear(64,1)
    def forward(self,x):
        assert x.shape[-1]==17
        z,_=self.lstm(x);return self.head(z[:,-1]).squeeze(-1)
def checkpoint(path,model,opt,epoch,best,history):
    state=dict(model=model.state_dict(),optimizer=opt.state_dict(),epoch=epoch,best_validation=best,history=history,experiment_hash=DIGEST,python_rng=random.getstate(),numpy_rng=np.random.get_state(),torch_rng=torch.get_rng_state(),cuda_rng=torch.cuda.get_rng_state_all())
    tmp=path.with_suffix('.tmp');torch.save(state,tmp);tmp.replace(path)
def validate(model):
    model.eval();sums=np.zeros(1326);nums=np.zeros(1326)
    with torch.no_grad():
        for x,y,bi,t in val_loader:
            q=model(x.to(DEVICE,non_blocking=True)).cpu().numpy();b=bi.numpy();err=(q-y.numpy())**2/p['loss_scales'][b]**2
            sums+=np.bincount(b,weights=err,minlength=1326);nums+=np.bincount(b,minlength=1326)
    return float(np.mean(sums[nums>0]/nums[nums>0])),int((nums>0).sum())
```

## 4. Train and validate
Run all 20 epochs and select the best validation checkpoint.


```python
model=MeteorologyLSTM().to(DEVICE);opt=torch.optim.Adam(model.parameters(),lr=.001)
assert model.lstm.input_size==17 and sum(v.numel() for v in model.parameters())==21313
x,y,bi,t=next(iter(loader));q=model(x[:4].to(DEVICE));scale=torch.tensor(p['loss_scales'][bi[:4].numpy()],dtype=torch.float32,device=DEVICE)
loss=(((q-y[:4].to(DEVICE))/scale)**2).mean();loss.backward();assert torch.isfinite(loss);opt.zero_grad();torch.cuda.synchronize()
first=0;best=float('inf');history=[]
if (OUT/'last.pt').exists():
    ck=torch.load(OUT/'last.pt',map_location="cpu",weights_only=False);assert ck['experiment_hash']==DIGEST,'Config/CSV changed; use another RUN_NAME.'
    model.load_state_dict(ck['model']);opt.load_state_dict(ck['optimizer']);first=ck['epoch'];best=ck['best_validation'];history=ck['history']
    random.setstate(ck['python_rng']);np.random.set_state(ck['numpy_rng']);torch.set_rng_state(ck['torch_rng'].cpu());torch.cuda.set_rng_state_all(ck['cuda_rng'])
started=time.time()
for epoch in range(first,20):
    model.train();sampler.epoch=epoch;total=0.;count=0
    for step,(x,y,bi,t) in enumerate(loader,1):
        opt.zero_grad(set_to_none=True);q=model(x.to(DEVICE,non_blocking=True));target=y.to(DEVICE,non_blocking=True)
        scale=torch.tensor(p['loss_scales'][bi.numpy()],dtype=torch.float32,device=DEVICE);loss=(((q-target)/scale)**2).mean();assert torch.isfinite(loss)
        loss.backward();nn.utils.clip_grad_norm_(model.parameters(),1.);opt.step();total+=float(loss.detach())*len(y);count+=len(y)
        if step%400==0:atomic(OUT/'status.json',dict(phase='training',epoch=epoch+1,batch=step,batches=len(loader)));print('PROGRESS',epoch+1,step,flush=True)
    v,nval=validate(model);history.append(dict(epoch=epoch+1,training_loss=total/count,validation_loss=v,validation_donors=nval,elapsed_seconds=round(time.time()-started)))
    improved=v<best;best=min(best,v);checkpoint(OUT/'last.pt',model,opt,epoch+1,best,history)
    if improved:checkpoint(OUT/'best.pt',model,opt,epoch+1,best,history)
    atomic(OUT/'history.json',history);print('EPOCH',history[-1],flush=True)
ck=torch.load(OUT/'best.pt',map_location="cpu",weights_only=False);model.load_state_dict(ck['model']);model.eval();print('Best epoch:',ck['epoch'])
atomic(OUT/'run_config.json',dict(config=CFG,csv_sha256=CSV_HASH,model_inputs=FEATURES,static_features=STATIC_FEATURES,parameters=sum(v.numel() for v in model.parameters()),torch_version=torch.__version__,numpy_version=np.__version__,gpu=torch.cuda.get_device_name(0)))
```

## 5. Test seen and unseen catchments
Export daily predictions and one performance row for every basin.


```python
def scores(y,q):
    good=np.isfinite(y)&np.isfinite(q); y,q=y[good],q[good]
    if len(y)<2 or y.std()==0: return dict(n=len(y),NSE=np.nan,KGE=np.nan,bias_percent=np.nan)
    nse=1-np.sum((q-y)**2)/np.sum((y-y.mean())**2)
    corr=np.corrcoef(y,q)[0,1] if q.std()>0 else np.nan
    kge=1-np.sqrt((corr-1)**2+(q.std()/y.std()-1)**2+(q.mean()/y.mean()-1)**2) if y.mean()!=0 else np.nan
    return dict(n=len(y),NSE=float(nse),KGE=float(kge),bias_percent=float(100*(q.sum()-y.sum())/y.sum()) if y.sum()!=0 else np.nan)

def evaluate(rs,tag):
    folder=OUT/tag;folder.mkdir(exist_ok=True);metrics=[]
    for i,r in enumerate(rs):
        ds=Windows([r],p,'test');ts=[];pred=[]
        with torch.no_grad():
            for x,y,bi,t in DataLoader(ds,batch_size=256,num_workers=0,pin_memory=True):
                q=model(x.to(DEVICE,non_blocking=True)).cpu().numpy()*p['ystd']+p['ymean'];ts.extend(t.numpy().tolist());pred.extend(q.tolist())
        ts=np.array(ts,dtype=int);q=np.array(pred);obs=r['y'][ts]
        assert np.isfinite(q).all() and np.all(r['dates'][ts]>=np.datetime64('2005-01-01')) and np.all(r['dates'][ts]<np.datetime64('2010-01-01'))
        save_csv(folder/(r['id']+'_predictions.csv'),[dict(date=str(r['dates'][t]),observed_mm_day=float(o),predicted_mm_day=float(v)) for t,o,v in zip(ts,obs,q)],['date','observed_mm_day','predicted_mm_day'])
        good=np.isfinite(obs)&np.isfinite(q);error=q[good]-obs[good]
        extra=dict(RMSE=float(np.sqrt(np.mean(error**2))) if len(error) else np.nan,MAE=float(np.mean(np.abs(error))) if len(error) else np.nan)
        metrics.append(dict(basin_id=r['id'],**scores(obs,q),**extra))
        if (i+1)%100==0 or i+1==len(rs):print('TEST',tag,i+1,'/',len(rs),flush=True);atomic(OUT/'status.json',dict(phase='testing',test=tag,completed=i+1,total=len(rs)))
    save_csv(folder/'metrics.csv',metrics);return metrics
unseen_metrics=evaluate(held,'unseen100');temporal_metrics=evaluate(donors,'donor_temporal')
all_metrics=[]
for tag,rows in [('Seen',temporal_metrics),('Unseen',unseen_metrics)]:
    for r in rows:
        r=dict(r,seen_unseen=tag,catchment_split='training_donor' if tag=='Seen' else 'unseen_test',test_start='2005-01-01',test_end='2009-12-31',model='Daymet + PCA10 means only')
        r['status']='scored' if r['n']>=2 and np.isfinite(r['NSE']) else ('no_test_observations' if r['n']==0 else 'insufficient_or_constant_observations')
        all_metrics.append(r)
all_metrics.sort(key=lambda r:r['basin_id'])
assert len(all_metrics)==len({r['basin_id'] for r in all_metrics})==1426
assert sum(r['seen_unseen']=='Seen' for r in all_metrics)==1326
assert sum(r['seen_unseen']=='Unseen' for r in all_metrics)==100
save_csv(OUTPUT_BASE/'basin_model_performance.csv',all_metrics,['basin_id','seen_unseen','catchment_split','model','test_start','test_end','n','NSE','KGE','RMSE','MAE','bias_percent','status'])
summary_rows=[]
for tag in ('Seen','Unseen'):
    rows=[r for r in all_metrics if r['seen_unseen']==tag];scored=[r for r in rows if np.isfinite(r['NSE'])]
    summary_rows.append(dict(seen_unseen=tag,total_basins=len(rows),scored_basins=len(scored),median_NSE=float(np.median([r['NSE'] for r in scored])),median_KGE=float(np.nanmedian([r['KGE'] for r in scored])),median_bias_percent=float(np.nanmedian([r['bias_percent'] for r in scored]))))
save_csv(OUTPUT_BASE/'model_performance_summary.csv',summary_rows)
atomic(OUT/'summary.json',dict(model='Daymet + PCA10 means only',best_epoch=ck['epoch'],test_start='2005-01-01',test_end='2009-12-31',groups=summary_rows))
atomic(OUT/'status.json',dict(phase='complete',basins=1426,seen=1326,unseen=100))
print('Both test groups complete. Basin-level metrics exported beside this notebook.')
```

## 6. Independently verify and report performance
Recalculate every metric from daily predictions, export the summary and display the results.


```python
# Independently verify the exported daily predictions against every saved basin score.
import html
with (OUTPUT_BASE/'basin_model_performance.csv').open() as f:
    performance=list(csv.DictReader(f))
assert len(performance)==len({r['basin_id'] for r in performance})==1426
assert sum(r['seen_unseen']=='Seen' for r in performance)==1326
assert sum(r['seen_unseen']=='Unseen' for r in performance)==100
history=json.loads((OUT/'history.json').read_text())
assert [r['epoch'] for r in history]==list(range(1,21))
chosen=json.loads((OUT/'summary.json').read_text())['best_epoch']
assert chosen==min(history,key=lambda r:r['validation_loss'])['epoch']
verified=0
for row in performance:
    group='donor_temporal' if row['seen_unseen']=='Seen' else 'unseen100'
    with (OUT/group/(row['basin_id']+'_predictions.csv')).open() as f:daily=list(csv.DictReader(f))
    dates=[r['date'] for r in daily]
    assert len(dates)==len(set(dates)) and all('2005-01-01'<=d<='2009-12-31' for d in dates)
    y=np.array([r['observed_mm_day'] for r in daily],dtype='float32')
    q=np.array([r['predicted_mm_day'] for r in daily],dtype='float64')
    assert np.isfinite(q).all()
    good=np.isfinite(y);y=y[good];q=q[good];n=len(y);error=q-y
    values=dict(NSE=np.nan,KGE=np.nan,bias_percent=np.nan,RMSE=float(np.sqrt(np.mean(error**2))) if n else np.nan,MAE=float(np.mean(np.abs(error))) if n else np.nan)
    assert n==int(row['n'])
    if n>=2 and y.std()>0:
        values['NSE']=float(1-np.sum(error**2)/np.sum((y-y.mean())**2))
        correlation=np.corrcoef(y,q)[0,1] if q.std()>0 else np.nan
        values['KGE']=float(1-np.sqrt((correlation-1)**2+(q.std()/y.std()-1)**2+(q.mean()/y.mean()-1)**2)) if y.mean()!=0 else np.nan
        values['bias_percent']=float(100*(q.sum()-y.sum())/y.sum()) if y.sum()!=0 else np.nan
    for metric,value in values.items():
        assert np.allclose(value,float(row[metric]),rtol=2e-5,atol=2e-5,equal_nan=True),(row['basin_id'],metric)
    verified+=1
    if verified%200==0:print('METRICS VERIFIED',verified,flush=True)
summaries=[]
for group in ['Seen','Unseen']:
    selected=[r for r in performance if r['seen_unseen']==group]
    for metric in ['NSE','KGE','RMSE','MAE','bias_percent']:
        a=np.array([float(r[metric]) for r in selected]);a=a[np.isfinite(a)]
        summaries.append(dict(seen_unseen=group,metric=metric,total_basins=len(selected),scored_basins=len(a),mean=float(a.mean()) if len(a) else np.nan,median=float(np.median(a)) if len(a) else np.nan,q1=float(np.quantile(a,.25)) if len(a) else np.nan,q3=float(np.quantile(a,.75)) if len(a) else np.nan,negative_count=int(sum(a<0)) if metric in ['NSE','KGE'] else ''))
save_csv(OUTPUT_BASE/'model_performance_summary.csv',summaries)
def table(rows):
    keys=list(rows[0]);return '<div style="overflow:auto"><table style="border-collapse:collapse"><tr>'+''.join('<th style="padding:7px;border:1px solid #ddd">'+html.escape(k.replace('_',' '))+'</th>' for k in keys)+'</tr>'+''.join('<tr>'+''.join('<td style="padding:7px;border:1px solid #ddd">'+html.escape(f'{r[k]:.4f}' if isinstance(r[k],float) else str(r[k]))+'</td>' for k in keys)+'</tr>' for r in rows)+'</table></div>'
report='<html><head><meta charset="utf-8"></head><body style="font:15px sans-serif;margin:24px"><h1>Daymet + PCA10 means only</h1><p>Seed 42. All 20 epochs completed. Best validation checkpoint: epoch '+str(chosen)+'. Test period: 2005–2009. Seven meteorological forcings and ten PCA means; 17 inputs, 64 hidden units.</p><h2>Model performance</h2>'+table(summaries)+'<h2>First ten basin rows</h2>'+table(performance[:10])+'<p>All 1,426 basin rows are in basin_model_performance.csv, including Seen/Unseen labels and missing-score status. Verification independently recalculated metrics from daily predictions for every basin.</p></body></html>'
(OUTPUT_BASE/'model_report.html').write_text(report)
atomic(OUTPUT_BASE/'results_verification.json',dict(status='passed',basins=verified,all_daily_metrics_recomputed=True,epochs=20,seed=42,best_epoch=chosen,inputs=FEATURES+STATIC_FEATURES,parameters=sum(v.numel() for v in model.parameters())))
atomic(OUTPUT_BASE/'completion.json',dict(status='complete',model='Daymet + PCA10 means only',basins=1426,seen=1326,unseen=100,epochs=20,seed=42,best_epoch=chosen))
display(HTML('<h2>Test performance, 2005–2009</h2>'+table(summaries)))
display(HTML('<h2>First ten basin results</h2>'+table(performance[:10])))
print('COMPLETE: all 20 epochs, both test groups and all-basin metric verification. Only the notebook and CSV were required inputs.')
```
