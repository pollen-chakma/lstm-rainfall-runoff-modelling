# Daymet + PCA10 mean/std: H256, output dropout 0.2, adaptive training

## Two reusable input files
Keep this notebook and `daymet_pca10_discharge_1980_2009.csv` together. Select **Model A PyTorch (GPU)** and run cells in order. The CSV is the sole model dataset. An immutable seed-42 baseline benchmark table is embedded in the final comparison cell, so the comparison does not need the earlier experiment folder. Outputs are created under `outputs/`. Copy the two inputs into an empty folder to train from scratch; rerunning here resumes matching epoch checkpoints.

## Fixed protocol and requested changes
Original 1,326 training donors and 100 unseen IDs; train 1980–1999, validate 2000–2004, test 2005–2009. Identical seven Daymet forcings plus PC01–PC10 mean and std (27 channels), baseline preprocessing, loss, 365-day windows, training batch 128, validation/inference batch 256, 256 sampled windows per donor/epoch, validation cap 128 per donor, gradient clipping 1.0 and seed 42. One LSTM layer changes from 64 to 256 units. Dropout 0.2 is applied to the last LSTM output before the linear head (not recurrent dropout). Model parameter count is 292,097.

Adam starts at 0.001. After **three consecutive non-improving validation epochs**, halve learning rate, minimum 0.00001. This is implemented explicitly to avoid library patience off-by-one differences. Any strictly lower validation loss resets both improvement counters. An LR reduction resets only its three-epoch counter. Stop after ten consecutive non-improving epochs or fifty completed epochs. Select the strictly lowest validation loss; test scores never select settings/checkpoints. Epoch checkpoints retain model/optimizer, scheduler counters, early-stopping counters, history and Python/NumPy/PyTorch/CUDA RNG states. Float32 arithmetic and original prediction handling are retained; no test-discharge clipping.

## Flow errors: definition fixed before evaluation
After selecting the validation-best checkpoint, derive observed discharge Q10 and Q90 separately for each basin from finite observed/predicted pairs in 2005–2009. Low-flow days have observed Q ≤ Q10; high-flow days have Q ≥ Q90. Ties are included, so subsets can exceed 10%. Report subset sample count, RMSE, MAE, mean bias (mm/day) and volume bias (%). A zero observed subset sum gives undefined percent bias; absolute errors remain available. No valid pairs give missing thresholds/errors. These observed test quantiles are **evaluation-only diagnostics**, not training inputs or fitted scaling. The same definitions and dates are applied to the old seed-42 benchmark. They describe relative within-basin extremes, not absolute drought/flood classes.

## Comparison and limits
Report seen and unseen separately, all basin rows including missing-score cases, paired differences on common scored basins, negative NSE/KGE counts, full daily predictions, history and selected epoch. One seed tests the combined architecture/training change, not its individual causes or repeatability. Earlier geographical overlap limitations are inherited unchanged. The 100 unseen basins have already been inspected in previous experiments. GPU results can vary slightly across hardware/software.

## 5. Evaluate both seen and unseen catchments

```python
def scores(y,q):
    good=np.isfinite(y)&np.isfinite(q); y,q=y[good],q[good]
    if len(y)<2 or y.std()==0: return dict(n=len(y),NSE=np.nan,KGE=np.nan,bias_percent=np.nan)
    nse=1-np.sum((q-y)**2)/np.sum((y-y.mean())**2)
    corr=np.corrcoef(y,q)[0,1] if q.std()>0 else np.nan
    kge=1-np.sqrt((corr-1)**2+(q.std()/y.std()-1)**2+(q.mean()/y.mean()-1)**2) if y.mean()!=0 else np.nan
    return dict(n=len(y),NSE=float(nse),KGE=float(kge),bias_percent=float(100*(q.sum()-y.sum())/y.sum()) if y.sum()!=0 else np.nan)

def evaluate(rs,tag):
    folder=OUT/tag;folder.mkdir(exist_ok=True);metrics=[]
    for i,r in enumerate(rs):
        ds=Windows([r],p,'test');ts=[];pred=[]
        with torch.no_grad():
            for x,y,bi,t in DataLoader(ds,batch_size=256,num_workers=0,pin_memory=True):
                q=model(x.to(DEVICE,non_blocking=True)).cpu().numpy()*p['ystd']+p['ymean'];ts.extend(t.numpy().tolist());pred.extend(q.tolist())
        ts=np.array(ts,dtype=int);q=np.array(pred);obs=r['y'][ts]
        assert np.isfinite(q).all() and np.all(r['dates'][ts]>=np.datetime64('2005-01-01')) and np.all(r['dates'][ts]<np.datetime64('2010-01-01'))
        save_csv(folder/(r['id']+'_predictions.csv'),[dict(date=str(r['dates'][t]),observed_mm_day=float(o),predicted_mm_day=float(v)) for t,o,v in zip(ts,obs,q)],['date','observed_mm_day','predicted_mm_day'])
        good=np.isfinite(obs)&np.isfinite(q);error=q[good]-obs[good]
        extra=dict(RMSE=float(np.sqrt(np.mean(error**2))) if len(error) else np.nan,MAE=float(np.mean(np.abs(error))) if len(error) else np.nan)
        metrics.append(dict(basin_id=r['id'],**scores(obs,q),**extra))
        if (i+1)%100==0 or i+1==len(rs):print('TEST',tag,i+1,'/',len(rs),flush=True);atomic(OUT/'status.json',dict(phase='testing',test=tag,completed=i+1,total=len(rs)))
    save_csv(folder/'metrics.csv',metrics);return metrics
unseen_metrics=evaluate(held,'unseen100');temporal_metrics=evaluate(donors,'donor_temporal')
all_metrics=[]
for tag,rows in [('Seen',temporal_metrics),('Unseen',unseen_metrics)]:
    for r in rows:
        r=dict(r,seen_unseen=tag,catchment_split='training_donor' if tag=='Seen' else 'unseen_test',test_start='2005-01-01',test_end='2009-12-31',model='Daymet + PCA10 mean/std H256 dropout0.2 adaptive50')
        r['status']='scored' if r['n']>=2 and np.isfinite(r['NSE']) else ('no_test_observations' if r['n']==0 else 'insufficient_or_constant_observations')
        all_metrics.append(r)
all_metrics.sort(key=lambda r:r['basin_id'])
assert len(all_metrics)==len({r['basin_id'] for r in all_metrics})==1426
assert sum(r['seen_unseen']=='Seen' for r in all_metrics)==1326
assert sum(r['seen_unseen']=='Unseen' for r in all_metrics)==100
save_csv(OUTPUT_BASE/'basin_model_performance.csv',all_metrics,['basin_id','seen_unseen','catchment_split','model','test_start','test_end','n','NSE','KGE','RMSE','MAE','bias_percent','status'])
summary_rows=[]
for tag in ('Seen','Unseen'):
    rows=[r for r in all_metrics if r['seen_unseen']==tag];scored=[r for r in rows if np.isfinite(r['NSE'])]
    summary_rows.append(dict(seen_unseen=tag,total_basins=len(rows),scored_basins=len(scored),median_NSE=float(np.median([r['NSE'] for r in scored])),median_KGE=float(np.nanmedian([r['KGE'] for r in scored])),median_bias_percent=float(np.nanmedian([r['bias_percent'] for r in scored]))))
save_csv(OUTPUT_BASE/'model_performance_summary.csv',summary_rows)
atomic(OUT/'summary.json',dict(model='Daymet + PCA10 mean/std H256 dropout0.2 adaptive50',best_epoch=ck['epoch'],test_start='2005-01-01',test_end='2009-12-31',groups=summary_rows))
atomic(OUT/'status.json',dict(phase='complete',basins=1426,seen=1326,unseen=100))
print('Both test groups complete. Basin-level metrics exported beside this notebook.')
```
