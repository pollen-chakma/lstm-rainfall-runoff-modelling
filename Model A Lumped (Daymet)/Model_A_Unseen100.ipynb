## Resume after a session ends

The prepared data, fixed split and checkpoints are stored in your home directory. In a new compatible GPU session, use the existing Model_A/.venv/bin/python to run train_unseen.py --resume run_seed42/last.pt, with all original settings unchanged. Check training_process.json and status.json before launching to avoid duplicate runs. The last checkpoint contains model weights, optimizer state and random states at the last completed epoch.

This is a single-seed baseline. The exact split retains training basins that overlap held-out drainage areas; inspect shared_polygon_overlap_screen.csv when interpreting spatial independence. Climate proxies are not official climate classifications.
